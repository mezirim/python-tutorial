Broadcasting in numpy - this is simply when you change the elements of a matrix with one single element 

```python
#Broadcasting with +
L = np.round(10*np.random.rand(3,2))
L
```

```python
L = L + 5
L #This adds 5 to each elemrnt in the array
```

```python
#Broadcasting with -
L = L - 5
L
```

```python
#Broadcasting with *
L = L * 2
L
```

```python
#Broadcasting with /
L = L/3
L = np.round(L,2)
L
```
