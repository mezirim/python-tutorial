function and arguement continuation
* = this keyword allows you to pass multiple arguement in a function

```python
def list_of_fruits(*args):
    print("roberta")
    for x in args:
        print(x)

list_of_fruits("banana", "mango", "pawpaw", "melon", "grape")
```

multiple arguement in dictionaries: the keywword is **

```python
def biodata(name="roberta", age = 20, gender = "female"):
    for x in biodata:
        print(x)
```

```python
def biodata (**args):
    for x in args:
        print(x, args[x])

biodata(name="roberta", age=29, gender="female")
```
